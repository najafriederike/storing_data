## Judges' Accountability - Exploratory Notebook

```python
# Load packages
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
# Load clean dataset
q1q2_oblig_df = pd.read_csv('../data/clean/q1q2_oblig_df.csv')
q1q2_oblig_df
```

```python
q1q2_oblig_df.columns
```

```python
# Load V-Dem data

vdem_df = pd.read_csv('../data/raw/V-Dem-v15.csv')
vdem_df.head()
```

```python
# Countries in dataset 

countries_list = q1q2_oblig_df['country'].unique().tolist()
countries_list
```

```python
# Subset V-Dem dataset
vdem_judicial_df = vdem_df[['country_name', 'year', 'v2juaccnt', 'v2jucorrdc', 'v2juhcind', 'v2juncind']]

# Rename column
vdem_judicial_df = vdem_judicial_df.rename(columns={'country_name': 'country'})

# Rename 'Türkiye' to 'Turkey'
vdem_judicial_df['country'] = vdem_judicial_df['country'].replace('Türkiye', 'Turkey')

vdem_judicial_df_filtered = vdem_judicial_df[
    (vdem_judicial_df['year'] >= 2000) &
    (vdem_judicial_df['year'] <= 2022) &
    (vdem_judicial_df['country'].isin(countries_list))
]

vdem_judicial_df_filtered.reset_index(drop=True)
```

```python
# Summary statistics - V-Dem scores

vdem_judicial_df_filtered[['v2juaccnt', 'v2jucorrdc', 'v2juhcind', 'v2juncind']].describe()
```

```python
# Min-Max Normalization of V-Dem scores (V1)

from sklearn.preprocessing import MinMaxScaler


vdem_judicial_df_normalized = vdem_judicial_df_filtered.copy()
score_cols = ['v2juaccnt', 'v2jucorrdc', 'v2juhcind', 'v2juncind']

vdem_judicial_df_normalized[score_cols] = (vdem_judicial_df_normalized[score_cols] - vdem_judicial_df_normalized[score_cols].min()) / \
                 (vdem_judicial_df_normalized[score_cols].max() - vdem_judicial_df_normalized[score_cols].min())

vdem_judicial_df_normalized#.describe()
```

```python
# Summary statistics - normalized V-Dem scores

vdem_judicial_df_normalized[['v2juaccnt', 'v2jucorrdc', 'v2juhcind', 'v2juncind']].describe()
```

```python
# Min-Max Normalization of V-Dem scores (V2 -- to double-check)

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

vdem_judicial_df_normalizedx = vdem_judicial_df_filtered.copy()
score_cols = ['v2juaccnt', 'v2jucorrdc', 'v2juhcind', 'v2juncind']

vdem_judicial_df_normalizedx[score_cols] = scaler.fit_transform(vdem_judicial_df_normalizedx[score_cols])
vdem_judicial_df_normalizedx
```

```python
# Merging datasets -- Judges' obligations dataset and V-Dem scores normalized

q1q2_oblig_vdem_df = pd.merge(q1q2_oblig_df, vdem_judicial_df_normalized, on=['country', 'year'], how='inner')
q1q2_oblig_vdem_df
```

#### Ranking of Change

```python
#### Ranking of Change
q1q2_oblig_df_sorted = q1q2_oblig_df.sort_values(['country', 'year'])
#q1q2_oblig_df_sorted
```

```python
# Calculate the difference between years (i.e. 2001-2000; 2002-2001; etc. -> first value returns NA)
q1q2_oblig_df_sorted['yearly_change'] = q1q2_oblig_df_sorted.groupby('country')['q1q2_sub-index_oblig'].diff()
# check NA
q1q2_oblig_df_sorted['yearly_change'].isna().sum()
```

```python
# Calculate absolute value (-0.5 -> 0.5; +0.5 -> 0.5)
q1q2_oblig_df_sorted['abs_change'] = q1q2_oblig_df_sorted['yearly_change'].abs()
# check NA
q1q2_oblig_df_sorted['abs_change'].isna().sum()
```

```python
# Calculate sum of all absolute values
reform_intensity = q1q2_oblig_df_sorted.groupby('country')['abs_change'].sum()

# Create dataframe to display Total Reform Score 
reform_intensity = reform_intensity.reset_index()
reform_intensity = reform_intensity.rename(columns={'abs_change': 'total_reform_score'})
reform_intensity = reform_intensity.sort_values('total_reform_score', ascending=False)

# Display ranking
reform_intensity
```

```python
# Bar Chart on Absolute Reform Change per Country

plt.figure(figsize=(15, 10))
plt.bar(reform_intensity['country'], reform_intensity['total_reform_score'])
plt.xticks(rotation=90)
plt.xlabel('Country')
plt.ylabel('Total Reform Intensity')
plt.title('Total Absolute Reform Change by Country')
plt.tight_layout()

# Export figure
plt.savefig("../figures/bar-chart_obligations_absolute-reform-change.png", format="png", bbox_inches='tight', pad_inches=0.5, dpi=300)

# Display figure
plt.show()
```
